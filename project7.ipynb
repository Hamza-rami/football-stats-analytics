```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# import data
df = pd.read_csv("players_data-2024_2025.csv")
```

```python
Comp = df['Comp'].value_counts().head(5).index
df_top5_comp = df[df['Comp'].isin(Comp)]
```

## Yellow cards leaders

```python
Yellow_card = df_top5_comp[['Player', 'Squad', 'CrdY']].sort_values(by='CrdY', ascending=False).head()
fig, ax = plt.subplots(1, 2, figsize=(18, 6))
ax[0].bar(Yellow_card['Player'], Yellow_card['CrdY'], edgecolor='black', color='Pink')
ax[0].grid()
ax[0].set_xlabel('Player')
ax[0].set_ylabel('Yellow Cards')
ax[0].set_title("Top 5 Goal Scorers")
```
