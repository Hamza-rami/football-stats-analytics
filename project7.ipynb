# 🟥 Discipline & Misc

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# import data
df = pd.read_csv("players_data-2024_2025.csv")
```

```python
Comp = df['Comp'].value_counts().head(5).index
df_top5_comp = df[df['Comp'].isin(Comp)]
```

## Yellow cards leaders

```python
Yellow_card_player = df_top5_comp[['Player', 'CrdY']].sort_values(by='CrdY', ascending=False).head(5)

Yellow_card_pos = df_top5_comp.groupby('Pos')['CrdY'].sum().sort_values(ascending=False).head(5)

fig, ax = plt.subplots(1, 2, figsize=(18, 6))

ax[0].bar(Yellow_card_player['Player'], Yellow_card_player['CrdY'], edgecolor='black', color='gold')
ax[0].set_xlabel('Player')
ax[0].set_ylabel('Yellow Cards')
ax[0].set_title("Top 5 Players with Most Yellow Cards")
ax[0].grid(axis='y', linestyle='--', alpha=0.7)

ax[1].bar(Yellow_card_pos.index, Yellow_card_pos.values, edgecolor='black', color='orange')
ax[1].set_xlabel('Position')
ax[1].set_ylabel('Yellow Cards')
ax[1].set_title("Top 5 Positions with Most Yellow Cards")
ax[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## Red cards leaders

```python
Yellow_card_player = df_top5_comp[['Player', 'CrdR']].sort_values(by='CrdR', ascending=False).head(5)

Yellow_card_pos = df_top5_comp.groupby('Pos')['CrdR'].sum().sort_values(ascending=False).head(5)

fig, ax = plt.subplots(1, 2, figsize=(18, 6))

ax[0].bar(Yellow_card_player['Player'], Yellow_card_player['CrdR'], edgecolor='black', color='red')
ax[0].set_xlabel('Player')
ax[0].set_ylabel('Yellow Cards')
ax[0].set_title("Top 5 Players with Most Yellow Cards")
ax[0].grid(axis='y', linestyle='--', alpha=0.7)

ax[1].bar(Yellow_card_pos.index, Yellow_card_pos.values, edgecolor='black', color='orange')
ax[1].set_xlabel('Position')
ax[1].set_ylabel('Yellow Cards')
ax[1].set_title("Top 5 Positions with Most Yellow Cards")
ax[1].grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## Penalties conceded

```python
pen_con = df_top5_comp[['Player', 'PKcon']].sort_values(by='PKcon', ascending=False).head(5)
plt.figure(figsize=(10,6))

plt.bar(pen_con['Player'], pen_con['PKcon'], edgecolor='black', color='orange')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
```

## Penalties won

```python
pen_won = df_top5_comp[['Player', 'PKwon']].sort_values(by='PKwon', ascending=False).head(5)
plt.figure(figsize=(10,6))

plt.bar(pen_won['Player'], pen_won['PKwon'], edgecolor='black', color='yellow')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
```
